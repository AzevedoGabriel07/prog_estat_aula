# Exercício 2

Considere $X\sim\operatorname{Beta}(2,5)$. Estime por Monte Carlo

$$
p=P(X>0{,}4).
$$

Calcule a variância do estimador e construa um intervalo de confiança de 95%.

**Valor teórico:** $p=0{,}23328$.

```python
def f_beta(x, alfa, beta):
  return math.gamma(alfa + beta)/(math.gamma(alfa)*math.gamma(beta)) * x**(alfa-1) * (1-x)**(beta-1)
```

```python
alpha = 2
beta = 5
amostra = []
n = 10000
ye = (alpha-1)/(alpha + beta - 2)
c = (ye**(alpha-1))*(1-ye)**(beta-1)*math.gamma(alpha + beta)/ (math.gamma(alpha)*math.gamma(beta))
for i in range(n):
  y = np.random.uniform()
  u = np.random.uniform()
  while u*c > f_beta(y, alpha, beta):
   y = np.random.uniform()
   u = np.random.uniform()
  amostra.append(y)

p = np.mean(np.array(amostra) > 0.4)
var = np.var(np.array(amostra) > 0.4)
ic_min , ic_max = p - (1.96*np.sqrt(var/n)), p + (1.96*np.sqrt(var/n))
print(f"O valor do estimador deve ser {p}")
print(f"O valor da variancia deve ser {var}")
print(f"O intervalo de confiança 95% vai de [{ic_min};{ic_max}]")
```

# Exercício 3

Considere a integral

$$
I=\frac{1}{\sqrt{8\pi}}\int_{-\infty}^{\infty}(x^2+2x+1)e^{-(x-1)^2/8}\,dx.
$$

Utilizando $X\sim N(1,4)$, reescreva a integral como uma esperança e estime $I$ por Monte Carlo. Calcule a variância do estimador e construa um intervalo de confiança de 95%.

**Valor teórico:** $I=8$.


```python
def normal(mi,sigma2):


    u1 = np.random.uniform()
    u2 = np.random.uniform()

    u_2pi = 2*np.pi*u1
    t = -np.log(u2)
    x = np.sqrt(2*t)*np.cos(u_2pi)

    return np.sqrt(sigma2)*x + mi
```

```python
mi = 1
sigma2 = 4

n = 100000
amostra = []
amostra_Var = [ ]
for i in range(n):
  valor = normal(mi,sigma2)
  amostra.append((valor+1)**2)
  amostra_Var.append(valor)



media = np.mean(amostra)
var = np.var(amostra_Var)

ic_min , ic_max = media - (1.96*np.sqrt(var/n)), media + (1.96*np.sqrt(var/n))
print(f"O valor do estimador deve ser {media}")
print(f"O valor da variancia deve ser {var}")
print(f"O intervalo de confiança 95% vai de [{ic_min};{ic_max}]")
```
