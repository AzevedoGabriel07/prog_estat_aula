```python
p_valores = np.linspace(0.01, 0.99, 10)
N = 1000
amostrinha_ing = []
amostra_ing = []
tempo_ing = []
for p in p_valores:
  inicio = time.time()
  for i in range(N):
    X = 1
    bern = 0
    while bern == 0:
      bern = (np.random.uniform()>p)
      X += 1
    amostrinha_ing.append(X)
  amostra_ing.append(amostrinha_ing)

  fim= time.time()
  tempo_ing.append(fim - inicio)
plt.plot(p_valores, tempo_ing, marker='o', label='Valores amostrais geo ingenua')
plt.grid(True)
plt.title('Comparação entre p e o tempo')
plt.legend()
plt.show()
```
